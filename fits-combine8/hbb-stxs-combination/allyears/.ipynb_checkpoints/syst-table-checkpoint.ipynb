```python
import ROOT as rt
import numpy as np
import os

from scipy import interpolate
import matplotlib.pyplot as plt

import pandas as pd
```

```python
systnames = ["AllExp","MCStat","DDB","JMSJMR","JESJER","TFMC",
             "AllTh","ggFtheory","VBFtheory","Vtheory"]

allsystnames = ["Total","AllStat"] + systnames

systtitles = {}
systtitles["Total"] = "Total"
systtitles["AllStat"] = "Statistical"
#systtitles["TFres"] = "QCD estimation ($F_{res}$)"
#systtitles["TQQ"] = "$t\bar{t}$ normalization & efficiency"
systtitles["AllExp"] = "Experimental"
systtitles["MCStat"] = "Size of simulated samples"
systtitles["DDB"] = "DDB tag efficiency"
systtitles["JMSJMR"] = "Jet mass scale & resolution"
systtitles["JESJER"] = "Jet energy scale & resolution"
systtitles["TFMC"] = "QCD estimation ($F_{P/F}$)"
systtitles["Other"] = "Other"
systtitles["AllTh"] = "Theoretical"
systtitles["ggFtheory"] = "ggF modeling"
systtitles["VBFtheory"] = "VBF modeling"
systtitles["VHttHtheory"] = "Other Higgs modeling"
systtitles["Vtheory"]= "V+jets modeling"
```

```python
def read_from_graph(poi, filename):
    
    if not os.path.isfile(filename):
        print("Missing "+filename)
        return [0,0]
    
    f = rt.TFile(filename)    
    t = f.Get("limit")
    
    npoints = t.GetEntries()
    
    mu_pos = []
    deltaNLL_pos = []
    
    mu_neg = []
    deltaNLL_neg = []
    
    center = 1
    i = 0
    
    for ievt in t:

        if "VBF" in poi:
            mu = t.rVBF
        elif "ggF" in poi:
            mu = t.rggF
            
        dnll = t.deltaNLL
        
        if i == 0:
            center = mu
            print(center)

        if mu < center:
            mu_neg += [mu]
            deltaNLL_neg += [dnll] 

        if mu > center:
            mu_pos += [mu]
            deltaNLL_pos += [dnll] 
            
        i += 1
            
    i_pos = np.argsort(mu_pos)
    i_neg = np.argsort(mu_neg)
            
    plt.scatter(x=mu_neg+mu_pos,y=deltaNLL_neg+deltaNLL_pos,label=filename,marker='.')
    plt.plot(mu_neg+mu_pos,np.ones(len(mu_neg+mu_pos)),color='black',linestyle=':')
    plt.legend()
    plt.show()
    
    mu_pos_sorted = [mu_pos[i] for i in i_pos]
    deltaNLL_pos_sorted = [deltaNLL_pos[i] for i in i_pos]
    
    mu_neg_sorted = [mu_neg[i] for i in i_neg]
    deltaNLL_neg_sorted = [deltaNLL_neg[i] for i in i_neg]
            
    #print('pos',deltaNLL_pos)
    f_pos = interpolate.interp1d(deltaNLL_pos_sorted, mu_pos_sorted)
    #print('neg',deltaNLL_neg)
    f_neg = interpolate.interp1d(deltaNLL_neg_sorted, mu_neg_sorted)
    
    print(f_neg(1)/np.sqrt(2)-center,f_pos(1)-center)
    
    return [f_neg(1)-center,f_pos(1)-center]
```

```python
def get_table(poi):

    table = pd.DataFrame()

    unc_total = read_from_graph(poi,"higgsCombine"+poi+".Total.root")
    table = pd.concat([table,pd.DataFrame([["Total"] + unc_total])])

    unc_allstat = read_from_graph(poi,"higgsCombine"+poi+"Stat.Total.root")
    table = pd.concat([table,pd.DataFrame([["Stat"] + unc_allstat])])

    for sys in systnames:
        
        filename = "higgsCombine_"+poi+sys+".Total.root"

        unc = read_from_graph(poi, filename)

        delta_mu_m = np.sqrt(unc_total[0]*unc_total[0] - unc[0]*unc[0])
        delta_mu_p = np.sqrt(unc_total[1]*unc_total[1] - unc[1]*unc[1])

        table = pd.concat([table,pd.DataFrame([[sys,delta_mu_m, delta_mu_p]])])

    return table

```

```python
vbftable = get_table("rVBF")
```
