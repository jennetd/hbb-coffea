```python
import uproot 
import mplhep as hep
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = uproot.open('templates/2016/TnPtemplates.root')
```

```python
f.keys()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for reg, ax in zip(['pass', 'fail'], [ax1, ax2]):
    plt.sca(ax)
    hs = []
    for key in ["un","match"]:
        h, bins = f['{}_{}_nominal'.format(reg,key)].to_numpy()
        hs.append(h)
        
    data = f['{}_muondata_nominal'.format(reg)].to_numpy()
    plt.errorbar(data[1][:-1], data[0], fmt='o', c='k')

    hep.histplot(hs, bins, stack=True, histtype='fill')
```
