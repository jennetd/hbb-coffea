```python
import uproot 
import mplhep as hep
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = uproot.open('2016-FitSingle/fitDiagnosticsTest.root')
```

```python
f['shapes_fit_s/wsfSingleFail'].keys()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for reg, ax in zip(['Pass', 'Fail'], [ax1, ax2]):
    plt.sca(ax)
    hs = []
    for key in "qcd", "wqq":
        h, bins = f['shapes_fit_s/wsfSingle{}/{}'.format(reg, key)].to_numpy()
        hs.append(h)
        
    data = f['shapes_fit_s/wsfSingle{}/data'.format(reg)]
    plt.errorbar(data.values(0), data.values(1), fmt='o', c='k')

    hep.histplot(hs, bins, stack=True, histtype='fill')
```
